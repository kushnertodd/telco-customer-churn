# Telco Customer Churn Analysis

## Telecommunications customer churn
Telecommunications companies provide customers a variety of services such as cell phone, landline, internet, and cable channels. 
[Customer churn](https://www.ibm.com/think/topics/customer-churn) is the number of existing customers lost over a given period of time. 
Telecommunications company customer churn rates vary widely depending on the service, but sample annual rates for wireless phone service (which makes up [45%](https://market.us/report/telecommunication-services-market/) of the market) can range from [10%](https://qz.com/att-q2-2026-earnings-subscriber-growth-revenue-072226) to over [40%](https://www.verizon.com/about/sites/default/files/2026-03/vz_4q25_foi-rc_031326-rem.pdf).
It costs more to recruit a new customer than it is to retain an existing customer, for example replacing the value of one lost customer can require signing up [three](https://www.mckinsey.com/capabilities/growth-marketing-and-sales/our-insights/experience-led-growth-a-new-way-to-create-value) to [five](https://propellocloud.com/blog/customer-retention-strategies-in-telecom-industry/) new customers. 

[Retaining customers](https://www.ibm.com/think/topics/customer-retention) has become harder as the cost of switching brands or providers has become cheaper.
Prioritizing [customer engagement](https://www.ibm.com/think/topics/customer-engagement), or providing positive experiences to customers, is an important protection against customer churn.
Customer churn analysis can help understand of how to identify potential changes to retain customers.
For example:
- [88% of customers](https://www.mckinsey.com/~/media/mckinsey/email/rethink/2023/08/2023-08-16c.html) say that good [customer service](https://www.ibm.com/think/topics/customer-service) was as important as the actual products or services that are purchased in making customers happy.
- [80% of customers](https://www.salesforce.com/service/digital-customer-engagement-platform/what-is-customer-engagement/) value the [positive experience](https://www.ibm.com/think/topics/customer-engagement) they have interacting with a company.
- [77% of consumers](https://www.ibm.com/thought-leadership/institute-business-value/report/ai-retail?utm_medium=Email&utm_source=External&utm_content=IBVWW&utm_term=20MFX&utm_id=1200-revolutionize-retail-AI-2024-press) make decisions about keeping a service when prices are too high.  

Factors influencing customer churn include:
- Customer satisfaction level
- Service costs
- The range of service options available
- Customer service experience
- Effort interacting with the company
- How likely they are to recommend the company and its products
- Demographic characteristics of the customers

This notebook performs machine learning analysis for a prepared telecommunications customer churn dataset. 
This notebook and data can be found in a [Github repository](https://github.com/kushnertodd/telco-customer-churn).

## Churn classification
We train three models:

- Logistic Regression — fast, interpretable baseline
- Random Forest — captures non-linear interactions
- XGBoost — production-grade gradient boosting; typically best on tabular data

### Logistic Regression

```python
lr = LogisticRegression(max_iter=3000, C=0.5, random_state=42)

lr.fit(X_train,y_train)
lr_pred = lr.predict(X_test)
lr_proba = lr.predict_proba(X_test)[:, 1]
roc_auc = roc_auc_score(y_test, lr_proba),
print(classification_report(y_test, lr_pred))
print(f'{'Logistic Regression':<25} ROC-AUC: {roc_auc[0]:.4f}')
cm = confusion_matrix(y_test, lr_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap=plt.cm.Blues)
plt.show()
```

### Random Forest

```python
rf = RandomForestClassifier(n_estimators=200, max_depth=8, random_state=42)
rf.fit(X_train,y_train)
rf_pred = rf.predict(X_test)
rf_proba = rf.predict_proba(X_test)[:, 1]
roc_auc = roc_auc_score(y_test, rf_proba),
print(classification_report(y_test, rf_pred))
print(f'{'Random Forest':<25} ROC-AUC: {roc_auc[0]:.4f}')
cm = confusion_matrix(y_test, rf_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap=plt.cm.Blues)
plt.show()
```

### XGBoost

```python
xgb = XGBClassifier(n_estimators=300, 
                    max_depth=4, 
                    learning_rate=0.05,
                    subsample=0.8, 
                    colsample_bytree=0.8,
                    eval_metric='logloss', 
                    random_state=42,
                    )
xgb.fit(X_train,y_train)
xgb_pred = lr.predict(X_test)
xgb_proba = xgb.predict_proba(X_test)[:, 1]
roc_auc = roc_auc_score(y_test, xgb_proba),
print(classification_report(y_test, xgb_pred))
print(f'{'XGBoost':<25} ROC-AUC: {roc_auc[0]:.4f}')
cm = confusion_matrix(y_test, rf_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap=plt.cm.Blues)
plt.show()
```
