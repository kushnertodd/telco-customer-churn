# Telco Customer Churn Analysis

## Telecommunications customer churn
Telecommunications companies provide customers a variety of [services](https://en.wikipedia.org/wiki/Supplier_convergence) such as cell phone, landline, internet, and cable channels. 
[Customer churn](https://www.ibm.com/think/topics/customer-churn) is the number of existing customers lost over a given period of time. 
Telecommunications companies can have [customer churn rates](https://customergauge.com/blog/average-churn-rate-by-industry) of around 31%. 
It costs more to recruit a new customer than it is to sell to an existing customer, replacing the value of one lost customer can require acquiring of [three new customers](https://www.mckinsey.com/capabilities/growth-marketing-and-sales/our-insights/experience-led-growth-a-new-way-to-create-value). 

[Retaining customers](https://www.ibm.com/think/topics/customer-retention) has become harder as the cost of switching brands or providers has become cheaper.
Prioritizing [customer engagement](https://www.ibm.com/think/topics/customer-engagement), or providing positive experiences to customers, is an important protection against customer churn.
Customer churn analysis can help understand of how to identify potential changes to retain customers.
For example:
- [88% of customers](https://www.mckinsey.com/~/media/mckinsey/email/rethink/2023/08/2023-08-16c.html) say that good [customer service](https://www.ibm.com/think/topics/customer-service) was as important as the actual products or services that are purchased in making customers happy.
- [80% of customers](https://www.salesforce.com/service/digital-customer-engagement-platform/what-is-customer-engagement/) value the [positive experience](https://www.ibm.com/think/topics/customer-engagement) they have interacting with a company.
- [77% of consumers](https://www.ibm.com/thought-leadership/institute-business-value/report/ai-retail?utm_medium=Email&utm_source=External&utm_content=IBVWW&utm_term=20MFX&utm_id=1200-revolutionize-retail-AI-2024-press) make decisions about keeping a service when prices are too high.  

Factors influencing customer churn include:
- Customer satisfaction level
- Service costs
- The range of service options available
- Customer service experience
- Effort interacting with the company
- How likely they are to recommend the company and its products
- Demographic characteristics of the customers

This notebook performs machine learning analysis and modeling for a prepared [telecommunications customer churn dataset](https://community.ibm.com/community/user/blogs/steven-macko/2019/07/11/telco-customer-churn-1113)
included with the [IBM Cognos Analytics](https://www.ibm.com/products/cognos-analytics) product.

### Review Data Types

The dataset contains a mix of numerical and categorical variables.
Many categorical features are currently stored as object types.
Detailed encoding and transformations will be handled during the feature engineering phase.

```python
df.info()
```

TotalCharges column type is unexpected, it appears as non-numeric (object) but is actually floating. It means at least one row has a non-numeric value (blanks in this case). Convert rows to numeric, then blanks become NaN (missing value). 

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors='coerce')
print("TotalCharges data type: ",df['TotalCharges'].dtype)
print(df.loc[df['customerID'] == '4472-LVYGI', ['customerID', "TotalCharges"]])
```

Similarly, the SeniorCitizen column is numeric but is actually categorical, with values 0/1 representing whether the customer is a senior or not.
To conform with other binary fields, convert 0/1 to 'No'/'Yes'.

```python
df['SeniorCitizen'] = df['SeniorCitizen'].map({0: 'No', 1: 'Yes'})
print("SeniorCitizen data type: ",df['SeniorCitizen'].dtype)
print(df_original.loc[0:5, ['SeniorCitizen']])
print(df.loc[0:5, ['SeniorCitizen']])
```

Remove customerID feature (unneeded).

```python
df = df.drop(columns=['customerID'])
```

Extract numeric and categorial column names.

```python
numeric_features = [feature for feature in df.columns if df[feature].dtype != 'O']
categorical_features = [feature for feature in df.columns if df[feature].dtype == 'O']
print('We have {} numerical features : {}'.format(len(numeric_features), numeric_features))
print('\nWe have {} categorical features : {}'.format(len(categorical_features), categorical_features))
```

Check missing values.
